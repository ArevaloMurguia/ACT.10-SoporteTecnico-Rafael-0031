```python
# UIII Act 10
# IA con Lista de Diccionarios
# Tema: Soporte Técnico
# Rafael Arevalo

equipos = []       # Lista vacía
losequipos = {}    # Diccionario vacío

print("====================================")
print("   IA DE SOPORTE TECNICO")
print("====================================")

while True:

    print("\n------------- MENU -------------")
    print("| 1 | Registrar equipo         |")
    print("| 2 | Revisar equipos          |")
    print("| 3 | Diagnosticar problema    |")
    print("| 4 | Mostrar diccionario      |")
    print("| 5 | Salir                    |")
    print("--------------------------------")

    op = input("Selecciona una opcion: ")

    # Opción 1
    if op == "1":

        print("\n--- REGISTRAR EQUIPO ---")
        nombre = input("Nombre del equipo: ")
        problema = input("Problema detectado: ")

        losequipos = {
            "nombre": nombre,
            "problema": problema
        }

        equipos.append(losequipos)

        print("Equipo registrado correctamente")

    # Opción 2
    elif op == "2":

        print("\n--- REVISAR EQUIPOS ---")

        if equipos == []:
            print("No hay equipos registrados")
        else:
            print(equipos)

    # Opción 3
    elif op == "3":

        print("\n--- DIAGNOSTICO IA ---")
        falla = input("Problema (lento/no enciende/sin internet): ")

        if falla == "lento":
            print("IA: Limpiar archivos temporales")
        elif falla == "no enciende":
            print("IA: Revisar cable de energia")
        elif falla == "sin internet":
            print("IA: Reiniciar modem")
        else:
            print("IA: Problema no identificado")

    # Opción 4
    elif op == "4":

        print("\n--- DICCIONARIO ACTUAL ---")

        if losequipos == {}:
            print("Diccionario vacio")
        else:
            print(losequipos)

    # Opción 5
    elif op == "5":

        print("\nPrograma terminado")
        break

    # Error
    else:
        print("\nOpcion incorrecta")
```
