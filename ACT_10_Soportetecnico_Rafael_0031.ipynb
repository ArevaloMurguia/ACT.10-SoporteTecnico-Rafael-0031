```python
# UIII Act 10
# IA con Lista de Diccionarios
# Tema: SoporteTecnico
# Arevalo Rafael
# Fecha: 25/May/26
# NL = 3

soportetecnico = []          # Lista vacía
lossoportestecnicos = {}     # Diccionario vacío

print("================================")
print("   IA DE SOPORTE TECNICO")
print("================================")

while True:

    print("\n----------- MENU -----------")
    print("1. Registrar soporte")
    print("2. Revisar soportes")
    print("3. Diagnosticar problema")
    print("4. Mostrar diccionario")
    print("5. Salir")
    print("----------------------------")

    op = input("Selecciona una opcion: ")

    if op == "1":

        print("\n--- REGISTRAR SOPORTE ---")
        nombre = input("Nombre del equipo: ")
        problema = input("Problema detectado: ")

        lossoportestecnicos = {
            "nombre": nombre,
            "problema": problema
        }

        soportetecnico.append(lossoportestecnicos)

        print("Soporte registrado correctamente")

    elif op == "2":

        print("\n--- REVISAR SOPORTES ---")

        if soportetecnico == []:
            print("No hay soportes registrados")
        else:
            print(soportetecnico)

    elif op == "3":

        print("\n--- DIAGNOSTICO IA ---")
        falla = input("Problema (lento/virus/no enciende): ")

        if falla == "lento":
            print("IA: Limpiar archivos temporales")

        elif falla == "virus":
            print("IA: Instalar antivirus")

        elif falla == "no enciende":
            print("IA: Revisar cargador")

        else:
            print("IA: Problema no identificado")

    elif op == "4":

        print("\n--- DICCIONARIO ACTUAL ---")

        if lossoportestecnicos == {}:
            print("Diccionario vacio")
        else:
            print(lossoportestecnicos)

    elif op == "5":

        print("\nPrograma terminado")
        break

    else:
        print("\nOpcion incorrecta")
```
